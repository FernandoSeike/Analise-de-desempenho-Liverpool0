```python
import pandas as pd
import numpy as np  

premier_league = pd.read_csv('England CSV.csv')

premier_league.head()

```

```python
duplicados = premier_league.duplicated().sum()
print(f"Número de linhas duplicadas: {duplicados}")
```

```python
print("Nomes das colunas no ficheiro:")
print(premier_league.columns.tolist())
print("-" * 50)
display(premier_league.head(3))
```

```python
jogos_liverpool = premier_league[premier_league['HomeTeam'] == 'Liverpool']
print(f"Número de jogos do Liverpool: {len(jogos_liverpool)}")
display(jogos_liverpool.head())
```

```python
colunas_essenciais = [
    'Date', 'HomeTeam', 'AwayTeam', 
    'FTH Goals', 'FTA Goals', 'FT Result', 
    'H Corners', 'A Corners', 
    'H Fouls', 'A Fouls', 
    'H Yellow', 'A Yellow', 'H Red', 'A Red'
]

jogos_liverpool = jogos_liverpool[colunas_essenciais].copy()

jogos_liverpool = jogos_liverpool.reset_index(drop=True)

print("Valores nulos encontrados em cada coluna:")
print(jogos_liverpool.isnull().sum())

print("\n--- TABELA LIMPA E PRONTA ---")
display(jogos_liverpool.head())
```

```python
# 1. Remover todas as linhas que tenham estatísticas em falta
jogos_liverpool = jogos_liverpool.dropna()

# 2. Reorganizar o índice novamente (pois acabámos de apagar 137 linhas)
jogos_liverpool = jogos_liverpool.reset_index(drop=True)

# 3. Confirmar o novo tamanho da nossa tabela e se os nulos desapareceram
print(f"Total de jogos do Liverpool 100% completos para análise: {len(jogos_liverpool)}")
print("\n--- Verificação Final de Nulos ---")
print(jogos_liverpool.isnull().sum())
```

```python

# 1. Golos: Quantos o Liverpool marcou e sofreu?
jogos_liverpool['Golos_Marcados'] = np.where(jogos_liverpool['HomeTeam'] == 'Liverpool', jogos_liverpool['FTH Goals'], jogos_liverpool['FTA Goals'])
jogos_liverpool['Golos_Sofridos'] = np.where(jogos_liverpool['HomeTeam'] == 'Liverpool', jogos_liverpool['FTA Goals'], jogos_liverpool['FTH Goals'])

# 2. Cantos: Pressão ofensiva do Liverpool
jogos_liverpool['Cantos_Favor'] = np.where(jogos_liverpool['HomeTeam'] == 'Liverpool', jogos_liverpool['H Corners'], jogos_liverpool['A Corners'])

# 3. Faltas: Agressividade do Liverpool
jogos_liverpool['Faltas_Cometidas'] = np.where(jogos_liverpool['HomeTeam'] == 'Liverpool', jogos_liverpool['H Fouls'], jogos_liverpool['A Fouls'])

# 4. Selecionar apenas as colunas essenciais para vermos o resultado final
colunas_finais = ['Date', 'HomeTeam', 'AwayTeam', 'Golos_Marcados', 'Golos_Sofridos', 'Cantos_Favor', 'Faltas_Cometidas']
display(jogos_liverpool[colunas_finais].head())
```

```python
# Calcular as estatísticas médias do Liverpool por jogo
medias_liverpool = jogos_liverpool[['Golos_Marcados', 'Golos_Sofridos', 'Cantos_Favor', 'Faltas_Cometidas']].mean()

# Formatar o resultado para ficar bonito e fácil de ler
print("--- ESTATÍSTICAS MÉDIAS DO LIVERPOOL POR JOGO ---")
print(round(medias_liverpool, 2))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Calcular o número total de jogos que estamos a analisar
total_jogos = len(jogos_liverpool)

plt.figure(figsize=(10, 6))
sns.set_theme(style="whitegrid")

cores = ['#C8102E', '#555555', '#C8102E', '#555555'] 

grafico = sns.barplot(x=medias_liverpool.index, y=medias_liverpool.values, palette=cores, hue=medias_liverpool.index, legend=False)

# 2. Adicionar o número total de jogos no título!
plt.title(f'Raio-X do Liverpool: Médias na Premier League\n(Amostra histórica: {total_jogos} jogos analisados)', 
          fontsize=16, fontweight='bold', pad=20)
plt.ylabel('Média por Jogo', fontsize=12)
plt.xlabel('')

for p in grafico.patches:
    grafico.annotate(format(p.get_height(), '.2f'), 
                     (p.get_x() + p.get_width() / 2., p.get_height()), 
                     ha = 'center', va = 'center', 
                     xytext = (0, 9), 
                     textcoords = 'offset points',
                     fontweight='bold', fontsize=11)

plt.xticks(ticks=[0, 1, 2, 3], labels=['Golos Marcados', 'Golos Sofridos', 'Cantos a Favor', 'Faltas Cometidas'], fontsize=11)

plt.tight_layout()
plt.show()
```

```python
# 1. Separar os jogos em casa e fora
jogos_casa = jogos_liverpool[jogos_liverpool['HomeTeam'] == 'Liverpool']
jogos_fora = jogos_liverpool[jogos_liverpool['AwayTeam'] == 'Liverpool']

# Contar quantos jogos o Python encontrou em cada lado
total_casa = len(jogos_casa)
total_fora = len(jogos_fora)

print(f"Raio-X de Jogos Encontrados: {total_casa} em Anfield | {total_fora} Fora de casa\n")

# 2. Calcular vitórias usando a matemática dos golos (Marcados > Sofridos)
vitorias_casa = len(jogos_casa[jogos_casa['Golos_Marcados'] > jogos_casa['Golos_Sofridos']])
vitorias_fora = len(jogos_fora[jogos_fora['Golos_Marcados'] > jogos_fora['Golos_Sofridos']])

# 3. Calcular a percentagem COM proteção contra erro de divisão por zero
taxa_vitoria_casa = (vitorias_casa / total_casa) * 100 if total_casa > 0 else 0
taxa_vitoria_fora = (vitorias_fora / total_fora) * 100 if total_fora > 0 else 0

# 4. Criar a tabela de dados para o gráfico
dados_vitorias = pd.DataFrame({
    'Cenário': ['Em Casa (Anfield)', 'Fora de Casa'],
    'Taxa de Vitória (%)': [taxa_vitoria_casa, taxa_vitoria_fora]
})

# 5. Desenhar o Gráfico de Barras Comparativo
plt.figure(figsize=(8, 6))
sns.set_theme(style="whitegrid")

cores = ['#C8102E', '#888888'] 
grafico_casa = sns.barplot(x='Cenário', y='Taxa de Vitória (%)', data=dados_vitorias, palette=cores, hue='Cenário', legend=False)

plt.title('O Fator Anfield: Impacto de Jogar em Casa nas Vitórias', fontsize=15, fontweight='bold', pad=20)
plt.ylabel('Taxa de Vitória (%)', fontsize=12)
plt.xlabel('')
plt.ylim(0, 100) # Fixar o eixo Y de 0 a 100%

# Colocar os números exatos no topo
for p in grafico_casa.patches:
    grafico_casa.annotate(f"{p.get_height():.1f}%", 
                     (p.get_x() + p.get_width() / 2., p.get_height()), 
                     ha = 'center', va = 'center', 
                     xytext = (0, 10), 
                     textcoords = 'offset points',
                     fontweight='bold', fontsize=13)
```

```python

# 1. Carregar a base de dados do zero para não haver falhas de memória
df_pl = pd.read_csv('England CSV.csv')

# 2. Filtrar o Liverpool e criar as colunas unificadas
jogos_liverpool = df_pl[(df_pl['HomeTeam'] == 'Liverpool') | (df_pl['AwayTeam'] == 'Liverpool')].copy()

jogos_liverpool['Golos_Marcados'] = np.where(jogos_liverpool['HomeTeam'] == 'Liverpool', jogos_liverpool['FTH Goals'], jogos_liverpool['FTA Goals'])
jogos_liverpool['Golos_Sofridos'] = np.where(jogos_liverpool['HomeTeam'] == 'Liverpool', jogos_liverpool['FTA Goals'], jogos_liverpool['FTH Goals'])

# 3. Separar os jogos em casa e fora
jogos_casa = jogos_liverpool[jogos_liverpool['HomeTeam'] == 'Liverpool']
jogos_fora = jogos_liverpool[jogos_liverpool['AwayTeam'] == 'Liverpool']

total_casa = len(jogos_casa)
total_fora = len(jogos_fora)

print(f"Sucesso! Encontrados: {total_casa} jogos em Anfield e {total_fora} jogos fora.\n")

# 4. Calcular vitórias e percentagens
vitorias_casa = len(jogos_casa[jogos_casa['Golos_Marcados'] > jogos_casa['Golos_Sofridos']])
vitorias_fora = len(jogos_fora[jogos_fora['Golos_Marcados'] > jogos_fora['Golos_Sofridos']])

taxa_vitoria_casa = (vitorias_casa / total_casa) * 100 if total_casa > 0 else 0
taxa_vitoria_fora = (vitorias_fora / total_fora) * 100 if total_fora > 0 else 0

# 5. Desenhar o gráfico final
dados_vitorias = pd.DataFrame({
    'Cenário': ['Em Casa (Anfield)', 'Fora de Casa'],
    'Taxa de Vitória (%)': [taxa_vitoria_casa, taxa_vitoria_fora]
})

plt.figure(figsize=(8, 6))
sns.set_theme(style="whitegrid")
cores = ['#C8102E', '#888888'] 

grafico = sns.barplot(x='Cenário', y='Taxa de Vitória (%)', data=dados_vitorias, palette=cores, hue='Cenário', legend=False)

plt.title(f'O Fator Anfield: Taxa de Vitória\n(Amostra de {total_casa + total_fora} jogos)', fontsize=15, fontweight='bold', pad=20)
plt.ylabel('Taxa de Vitória (%)', fontsize=12)
plt.xlabel('')
plt.ylim(0, 100)

for p in grafico.patches:
    grafico.annotate(f"{p.get_height():.1f}%", 
                     (p.get_x() + p.get_width() / 2., p.get_height()), 
                     ha = 'center', va = 'center', 
                     xytext = (0, 10), 
                     textcoords = 'offset points',
                     fontweight='bold', fontsize=13)

plt.tight_layout()
plt.show()
```
